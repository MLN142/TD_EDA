```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df=sns.load_dataset('titanic')
```

```python
print(df.shape)
print(df.columns)
```

```python
print(df.info())
```

```python
print(df.describe())
```

```python
print(df.isnull().sum())
```

```python
print(df.head(10))
```

```python
print(df["survived"].value_counts())
print(df["survived"].mean())
```

```python
sns.countplot(data=df,x="survived")
plt.title("Survivor Count")
plt.show()
```

```python
sns.countplot(data=df,x="sex")
plt.title("Gender Count")
plt.show()
```

```python
sns.countplot(data=df,x="survived",hue="sex")
plt.title("Survival on basis of gender")
plt.show()
```

```python
df["age_group"]=pd.cut(df["age"],bins=[0,18,35,65,100],labels=["Child","Young_Adult","Adult","Senior"])
sns.countplot(data=df,x="age_group",hue="survived")
plt.show()
```

```python
print(df.groupby("sex")["survived"].mean())
```

```python
print(df.groupby("class")["survived"].mean())
```

```python
sns.histplot(data=df,x="age",bins=80)
```

```python
sns.boxplot(data=df , x='survived',y="sex")
```
